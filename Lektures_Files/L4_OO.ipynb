## Encapsulation

Note: Often the terms "encapsulation" and "information hiding" are used interchangeably. (for those who are interested in more details, see e.g. the article [Encapsulation is not information hiding](https://www.infoworld.com/article/2075271/encapsulation-is-not-information-hiding.html)).

***Encapsulation* is the bundling of data and the methods that operate on that data.**
E.g.,




````python
class Employee:
  def __init__(self, firstname, lastname, persId):
    self.firstname = firstname
    self.lastname = lastname
    self.persId = persId
    self.skills = []
    self.salary = 50000

  def addSkill(self, skill):
    self.skills.append(skill)

  def prais(self):
    print("Great job, {}!".format(self.firstname))
````

## Information hiding

This term was introduced in 1972 by David Parnas. He advocated hiding "difficult design decisions or design decisions which are likely to change."

One technique that fosters information hiding is to adapt the visibility of attributes and methods. 
The two most used levels of visibility are "**public**" and "**private**". Additionally, depending on the programming language, there are other levels such as "**protected**", "package"-wide visibility etc.

For example, in statically typed languages such as Java, we would want to avoid that code in other parts of the program or third party code (generally referred to as 'client'-code) become invalid, if we decided to change a certain attribute in our class from type ArrayList to type LinkedList. Therefore, firstly, we try not to expose the exact type but rather a more general type (such as List). However, this is more a matter of abstraction not so much of visibility. Secondly, we might want to make the attribute not visible to client code (as to avoid its usage there alltogether). Consequently, we would decide to set its visibility to private.




In python, the underscore `_` or double-underscore `__` (called "dunder") is used to control the visibility of attributes and methods.

0) **PUBLIC**: In general, everything is public (see exception below)

1) Dunder-prefix AND dounder-suffix combined means something special in python (e.g,. `__init__`, `__str__`, ...). Do NOT use this for your stuff (typically).

2) **PROTECTED**: Underscore-prefix: e.g., `_myProtectedAttributeOrMethod`. This means one shouldn't access this method/attribute, as it is not part of the API (Application Programming Interface). Such a method/attribute is still visible outside of the class. This is a **convention only**.

3) **PRIVATE**: Dunder-prefix, e.g., `__myPrivateStuff`. Such attributes or methods are not visible/accessible outside the class. This is achieved by *name-mangling*. However, nothing is truely private in Python (see below).

````python
class MyClass:

  def __init__(self):  #special usage
    self.a = "a"       #public
    self._a = "_a"     #protected
    self.__a = "__a"   #private


  def test(self):
    print("{}, {}, {}".format(self.a, self._a, self.__a))

  def m(self):  #public
    print("m")

  def _m(self):  #protected
    print("_m")

  def __m(self):  #private
    print("__m")
````

````python
c = MyClass()

### Attributes

print(c.a)
print(c._a)
#print(c.__a)          #AttributeError: 'MyClass' object has no attribute '__a'
print(c._MyClass__a)   #we can use name mangling to access the private attribute. Don't do this!

print("--------")

c.m()
c._m()
#c.__m()                 #AttributeError: 'MyClass' object has no attribute '__m'
c._MyClass__m()          #we can use name mangling to call the private method. Don't do this!

vars(c)
````

## Class variables

As already seen, each instance of a class (e.g., `Employee` above) has its own namespace with its own **instance variables**. Two different instances of this class each have their own instance variables. Setting the name, for example, in one instance doesn’t affect the other instance.

In addition, a class can also have variables: **class variables**. They are set as part of the class definition.



````python
class Employee:

  availableSkills = ["Python", "C", "C#", "Java", "Lisp"]      #### this is a class variable

  def __init__(self, firstname, lastname, persId):
    self.firstname = firstname
    self.lastname = lastname
    self.persId = persId
    self.skills = []
    self.salary = 50000

  def addSkill(self, skill):
    if (skill in self.availableSkills):       ## accessing the class variable
      self.skills.append(skill)
    else:
      raise Exception("unknown skill")

  def prais(self):
    print("Great job, {}!".format(self.firstname))
````

````python
e1 = Employee("Maria", "Musterfrau", 1)
print(e1.skills)
e1.addSkill("C#")
print(e1.skills)
#e1.addSkill("Julia")  # will raise Exception: unknown skill (note: don't forget to re-execute the class definition above before you execute this cell)

e2 = Employee("Max", "Mustermann", 2)
print(Employee.availableSkills)
print(e2.availableSkills)

Employee.availableSkills.append("Rust")
e2.availableSkills.append("Julia")

print(Employee.availableSkills)
print(e2.availableSkills)

e1.addSkill("Julia")
print(e1.skills)

e1.firstName = "xxxx"     #Note: this is used only for the example below (firstname vs. firstName)
e1.prais()

#Note: running this code multiple times will lead to duplicates in class property 'availableSkills': e.g., "['Python', 'C', 'C#', 'Java', 'Lisp', 'Rust', 'Julia', 'Rust', 'Julia']"
#      re-run the class definition to start with the inital value
````

## **Getters and Setters** vs. **Properties**

````python
class P:
  def __init__(self, first, last, age):
    self.first = first
    self.last = last
    self.age = age
````

So far, we almost always used public attributes (instance variables) in classes. 

In some languages it would be a Nogo to have public instance variables like "first", "last", or "age" above. In Java, for example, one would make the instance variable private and provide a public method to read (get) the value and a public method to write (set) the value; these methods are called getter- and setter method, respectively.

jfyi: (no need to know/learn Java syntax in this lecture)
```
# //JAVA 

# //Java-Nogo:
public class P {
  public int age     //considered bad practice (public attributes)
}

P p = new P();
p.age = 5;


# //Java-WayToGo
public class P {
  private int age;
  public int getAge() {return age;}
  public void setAge(int a) {age = a;}
}
```

Why is this second approach better?
For example, at some point we realize that a negative age doesn't make any sense.
In the Nogo-case, we cannot avoid a (client-) code like this:
`myP.age = -5`

With the second code, we could simply modify the implementation of the setter method:
```
public void setAge(int a) {
  if a>0
    age = a;
  else
    //..
}
```

How can we achive the same thing in python?

Note: **In python, introducing setters/getters like in Java would be a Nogo!** (see below)

````python
#without getter/setter, the code is much more readable
p = P("first","last",20)
print(p.age)

#however, we can set the age to a negative value, which we want to avoid
p.age = -10
print(p.age)
````

````python
#We could avoid setting negative age values by using getter/setter methods ----   DON'T DO THIS !!   ;-)  
class P:
  def __init__(self, first, last, age):
    self.first = first
    self.last = last
    self._age = age

  def getAge(self):
    return self._age

  def setAge(self, a):
    if a <= 0 :
      raise Exception("illegal age")
    else:
      self._age = a

p = P("f","l", 10)
print(p.getAge()) 
#p.setAge(-5)   ## this raises an exception: -->   Exception: illegal age
print(p.getAge())

#Note that the client code is not as easy to read; also we have more code to write in the classes (imagine a class with many attributes)
````

### **--> the Python-way**: Property-Decorator     

(another use-case for decorators!)

````python
#the python-solution
#we start with this code   (public attributes, no getter/setter)
class P:
  def __init__(self, first, last, age):
    self.first = first
    self.last = last
    self.age = age

p = P("first","last",20)
p.age = -10
print(p.age)
````

````python
#now we realize, we want to add the (negativity-)check for the age attribute

#1) we make age private (or protected)
class P:
  def __init__(self, first, last, a):
    self.first = first
    self.last = last
    self._age = a

p = P("first","last",20)
p.age = -10
p.dsfsdfsfs = 77
print(p.age)
#p.__dict__
vars(p)

#Note, when setting the "age" attribute in the client, we add another attribute (in addition to the "_age" attribute).
#So far, this is not what we want!
````

````python
#The python-way:    Decorators!

class P:
  def __init__(self, first, last, a):
    self.first = first
    self.last = last
    self._age = a

  @property
  def age(self):
    print("GETTER!")
    return self._age

  @age.setter               #note the naming convention!
  def age(self, a):
    print("SETTER!")
    if (a<0):
     raise Exception("negative age") #print("ERROR. negative age")
    else:
      self._age = a

p = P("first","last",20)
try:
  p.age = -10     #note: not p.age(-10)
except Exception:
  print("hmmm, seems like I entered an illegal value")
print(p.age)
print("----")
p.age = 88
print(p.age)
````

### More Decorators 

### **`@classmethod`**  

Instead of the object (self) the uninstantiated class is passed as the first argument (cls)

````python
class Pers:
  
  availableSkills = []

  def __init__(self, firstname, lastname):
    self.firstname = firstname
    self.lastname = lastname

  @classmethod
  def fromString(cls, nameString):
    f,n = nameString.split()
    print("Instantiating class {}".format(cls))
    newPers = cls(f, n)
    return newPers

  @classmethod
  def check(cls, skill):
    return skill in cls.availableSkills

  def __str__(self):
    return self.firstname + " " + self.lastname
````

````python
p = Pers.fromString("Elvis P.")  # note only one argument
print(p.lastname)

Pers.check("C#")
````

### **`@staticmethod`**  

````python
class Pers:
  def __init__(self, firstname, lastname):
    self.firstname = firstname
    self.lastname = lastname

  @staticmethod
  def isValidName(nameString):
    return len(nameString.split()) == 2
````

````python
s = "Elvis P."
p = Pers("Elvis" , "P.")
print(Pers.isValidName(s)) #works independently of @staticmethod
print(p.isValidName(s)) #only works if isValidName is decorated with @staticmethod
````

# **Inheritance**

(see e.g., https://en.wikipedia.org/wiki/Inheritance_(object-oriented_programming))

````python
class Person:
  def __init__(self, firstname, lastname):
    self.firstname = firstname
    self.lastname = lastname

  def sayHi(self):
    print("Hi my name is {}".format(self.firstname))

  def __repr__(self):
    return "{} {}".format(self.firstname, self.lastname)

````

````python
p = Person("Maria", "Huber")
print(p)
````

````python
#we inherit from class person
class Student(Person):
  pass
````

````python
#when we want to create a Student, we get the following Error message: 
#TypeError: __init__() missing 2 required positional arguments: 'firstname' and 'lastname'
s = Student()
````

````python
class Student(Person):
  def __init__(self, firstname, lastname):
    super().__init__(firstname,lastname)
````

````python
#s = Student() # now we get the same error (but this time, it is clear why....)
s = Student("A", "BC")
#print(s)
s.sayHi() # we can call sayHi, because students inherit this method from persons
````

````python
print(s)
````

````python
dir(s)
````

````python
#so far class Student doesn't provide any benefit. We'll extend it:
#1) we add extra information
class Student(Person):
  def __init__(self, firstname, lastname, matnr):
    super().__init__(firstname,lastname)
    self.matnr = matnr
    self.courses = []

#2) we add extra methods
  def enroll(self,course):
    self.courses.append(course)

#3) overwrite inherited methods (sayHi has been defined already in Person)
  def sayHi(self):
    super().sayHi() #we may call the method of the base class
    print("  I'm enrolled in the following courses: {}".format(self.courses))
````

````python
s = Student("Max", "M", 12345)
s.enroll("Anw.Entw.Py")
s.sayHi()
````

### Polymorphism

Polymorphism means the ability to take various forms.
It is a very important concept in programming.

Depending on the actual implementation of a method (remember a method can be overwritten in subclasses), objects behave differently -- depending on their type.

````python
myList = [Person("Hans", "H."), Person("Maria", "M."), Student("Sandra","S.", 123), Person("Niki","N."), Student("Stefan", "S.",345)]
def greet(lst):
  for p in lst:
    p.sayHi()

greet(myList)
````
