# Higher-order functions

... a function that takes a function as parameter **OR** that returns a function 

the returned function can also be an inner function

````python
def hof():
    def inner():
        return 5
    
    return inner    ##very important: we return the function, not the result of invoking the function (there is no "()" after the function)

#test
v = hof()
print(v)
w = v()
print (w)





````

````python
def hof2(x):
    def inner():
        return x*10
    
    return inner

#test
v = hof2(5)
print(v)
w = v()
print(w)

## LAB: this output might not be a big surprise. However, there is something interesting about it ....... Can you spot it? 

#Hint: "closure"

````

````python
#Note: the inner function can also take arguments of course:

def hof3(x):
    def inner(y):
        return x*y
    
    return inner

#test
v = hof3(5)
print(v)
w = v(100)
print(w)
````

# Decorators

now we try out functions that get a **function as argument** **AND** also **return a function**

````python
#first some motivation:

def helperFunc():
    print("This is a PRINT statement")

def foo(bar):
    print("I will now call " + str(bar))
    bar()
    print("I just called " + str(bar))


foo(helperFunc)
````

````python
def helperFunc():
    print("This is a PRINT statement")

def decorate(bar):
    def foo():
        print("I will now call " + str(bar))
        bar()
        print("I just called " + str(bar))

    return foo   #we return the inner function

newFunc = decorate(helperFunc)

newFunc()

##what have we achieved?   ..... we can use decorate to alter the behavior of a function .... e.g., add additional print statements as in our case here
````

````python
def logBeforeAfterWrapper(f):
  def logBeforeAfter():
    print("START EXECUTION OF " + f.__name__)
    f()
    print("END EXECUTION OF " + f.__name__) 

  return logBeforeAfter             ##again: we return the function, not the result of invoking the function (there is no "()" after the function)

def myFunction():
  print("this is myFunc")

myFunction()
print(myFunction)
print('---------')

myFunction = logBeforeAfterWrapper(myFunction)   # we replace the original function with the wrapper

myFunction()

print('/////////////')

##In this case we REPLACED one function by its decorated version

````

````python
#another example

def plus(a,b):
  return a+b

def mult(a,b):
  return a*b

plus = mult

print(plus(3,4))
print(mult(3,4))
````

````python
#instead of writing   myFunction = logBeforeAfterWrapper(myFunction)    
#we can do the following:

@logBeforeAfterWrapper
def anotherFunc():
  print("this is another func")

anotherFunc()

#logBeforeAfterWrapper    is called a "decorator"!


````

A decorator is a function `d` that takes some function `f` and returns yet another function `w`. It is used to modify the behavior of function `f`.

````python
#here are two other decorators (doTwice, doNix):

def doTwice(f):
  def wrapper_twice():
    f()
    f()
  return wrapper_twice

def doNix(f):
  def wrapper():
    pass
  return wrapper

#@doNix
#@doTwice
def sayHi():
  print("Hi!")

sayHi()
````

# `*args and **kwargs`

.... allow you to pass multiple arguments or keyword arguments to a function

````python
#Examples:
def mysum(i1, i2):
  return i1+i2

print (mysum(2,3))

# :-(   this works, but only for 2 arguments, what if we wanted it to work with an arbitrary number of arguments
````

````python
def mysum(intlist):
  s = 0
  for i in intlist:
    s += i
  return s

print( mysum([1,2,3,4,5]) )

#ok, but this requires the integers be packed in a list. what if we don't want this?

#we could also use a tuple instead of a list. e.g., (1, 2, 3)
#--> a tuple supports iteration and slicing, but it is not mutable!

````

## `*args`



````python
def mysum(*args):
  s = 0
  for i in args:
    s += i
  return s

print (mysum(1,2,3,4,5) )

#when mysum is invoked, all the arguments get packed into a single iterable object called args (note: args is just a name, you can choose it to be anything else)

# *  is an unpacking operator
# the result of the unpacking operation is a tuple (not a list) 


````

## `**kwargs`

Instead of accepting positional arguments, it accepts keyword (also called 'named') arguments.


````python
def kwExample(**kwargs):
  print (kwargs)
  print (kwargs.keys())
  print (kwargs.values())
  for e in kwargs.items():
    print("{} = {}".format(e[0], e[1] ))

kwExample(key1 = "v1", key2 = "v2", key3 = "v3")
````

When writing your own functions with `*args and **kwargs`, you have to follow this ordering:


1.   standard arguments
2.   `*args` arguments
3.   `**kwargs` arguments


Example: 

```
def example(a, b, *args, **kwargs):
  pass
```
  


# Decorators with support for parameters

````python
#this is doTwice from above   (without parameters)
#
#def doTwice(f):
#  def wrapper_twice():
#    f()
#    f()
#  return wrapper_twice


def doTwiceParams(f):
  def wrapper_twice(*args):
    f(*args)
    f(*args)
  return wrapper_twice

@doTwiceParams
def sayHi(name):
  print("Hi {}!".format(name))


@doTwiceParams
def sayHiNoname():
  print("Hi")

  

sayHi("xxx")
sayHiNoname()
````

# Decorators and returned values

````python
print (sayHi("Hans"))

#sayHi returns None. This is ok for this case
````

````python
#what if sayHi returned a value?
@doTwiceParams
def sayHi(name):
  print("Hi {}!".format(name))
  return "Nice meeting you"

print (sayHi("Hans"))

#what happened to "Nice meeting you" ?
````

````python
def doTwiceParams(f):
  def wrapper_twice(*args, **kwargs):
    f(*args, **kwargs)
    return f(*args, **kwargs)     # we need to return the returned value of the last call to f
  return wrapper_twice

@doTwiceParams
def sayHi(name):
  print("Hi {}!".format(name))
  return "Nice meeting you"

print (sayHi("Hans"))
````
